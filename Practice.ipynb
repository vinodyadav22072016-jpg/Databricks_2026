```python
print("Hello World")

from pyspark.sql import functions as F
from pyspark.sql.window import Window

df = spark.createDataFrame([
    (101, "Amit", "2026-01-01"),
    (101, "Amit", "2026-02-01"),
    (102, "Priya", "2026-01-10")
], ["emp_id", "name", "updated_date"])

w = Window.partitionBy("emp_id").orderBy(
    F.col("updated_date").desc()
)

result = (df.withColumn("rn", F.row_number().over(w))
            .filter("rn = 1")
            .drop("rn"))

result.show()
```
